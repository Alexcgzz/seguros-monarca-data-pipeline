```python
def perfilar(df, nombre):
    print(f"======= {nombre} =======")
    print("Tamaño (filas, columnas):", df.shape)
    print("\nNulos por columna: ")
    print(df.isnull().sum())
    print("\nTipos de datos:")
    print(df.dtypes)
    print()
```

```python
perfilar(polizas, "POLIZAS")
perfilar(siniestros, "SINIESTROS")
perfilar(pagos, "PAGOS")
```
